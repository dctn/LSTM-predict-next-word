```python
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense
from tensorflow.keras.datasets import imdb
import tensorflow as tf
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)
```

```python
X_train.shape,y_train.shape
```

```python
sample_review = X_train[0]
sample_label = y_train[0]
sample_label
```

```python
data_index = imdb.get_word_index()
reversed_data_index = {value:key for key,value in data_index.items() }
reversed_data_index.get(150)
```

```python
" ".join([reversed_data_index.get(i-3,"?") for i in sample_review])
```

```python
max_len = 100
X_train= pad_sequences(X_train,maxlen=max_len)
X_test= pad_sequences(X_test,maxlen=max_len)
X_train[0]
```

## creating model

```python
max_features = 20000
model = Sequential()
model.add(Embedding(max_features,128,input_length=max_len))
model.add(SimpleRNN(128,activation="relu"))
model.add(Dense(1,activation="sigmoid"))
```

```python
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping
early_stopping = EarlyStopping(monitor="val_loss", patience=5,restore_best_weights=True)
```

```python
model.fit(
    X_train,
    y_train,
    batch_size=32,
    epochs=10,
    validation_data=(X_test,y_test),
    callbacks=[early_stopping],
    validation_split=0.2,
)
```

```python
model.save("own_model.keras")
```

## predictions

```python
def pre_process(review):
    review_words = review.lower().split()
    process_review = [data_index.get(word,2) for word in review_words]
    pad_seq = pad_sequences([process_review],maxlen=max_len)
    return pad_seq
```

```python
def predict_review(review):
    pre_processed = pre_process(review)
    prediction = model.predict(pre_processed)
    result = "Postive" if prediction >= 0.5 else "Negative"
    return result,prediction[0][0]
```

### recreating with Pytorch

```python
import torch.nn as nn
import torch
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
device
```

```python
class RNN(nn.Module):
    def __init__(self,embedding_size,hidden_state_size,max_features):
        super(RNN, self).__init__()
        self.embedding_layer = nn.Embedding(max_features, embedding_size)
        self.rnn = nn.RNN(input_size=embedding_size,hidden_size=hidden_state_size,
                          batch_first=True,nonlinearity="tanh")
        self.fc = nn.Linear(hidden_state_size,1)
        self.sigmoid_layer = nn.Sigmoid()

    def forward(self,x):
        x = self.embedding_layer(x)
        output, hidden = self.rnn(x)
        h_n = hidden.squeeze(0)
        out = self.fc(h_n)
        return out
```

```python
torch_rnn = RNN(embedding_size=1280,hidden_state_size=1280,max_features=20000).to(device)
```

```python
from torchinfo import summary
summary(
    torch_rnn,
    input_size=(2, 8),   # batch_size=2, seq_len=100
    dtypes=[torch.long]    # REQUIRED for Embedding
)
```

```python
X_train = torch.tensor(X_train, dtype=torch.long)
y_train = torch.tensor(y_train, dtype=torch.float32)

X_test = torch.tensor(X_test, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.float32)
```

```python
from torch.utils.data import DataLoader,TensorDataset

train_dataset = TensorDataset(X_train, y_train)
test_dataset = TensorDataset(X_test, y_test)

train_dataloader = DataLoader(train_dataset, batch_size=32,shuffle=True)
test_dataloader = DataLoader(test_dataset, batch_size=32)
```

```python
loss_func = nn.BCEWithLogitsLoss()
optim = torch.optim.Adam(torch_rnn.parameters(),lr=0.001)
```

```python
from tqdm.autonotebook import tqdm
EPOCHS = 5

for epoch in tqdm(range(EPOCHS)):
    # print(f"Epoch {epoch+1}/{EPOCHS}")
    torch_rnn.train()
    total_loss = 0

    for batch,(x,y) in enumerate(train_dataloader):
        x_batch = x.to(device)
        y_batch = y.to(device)

        y_logits = torch_rnn(x_batch).squeeze()
        loss = loss_func(y_logits,y_batch)
        optim.zero_grad()
        loss.backward()
        optim.step()

        total_loss += loss.item()
    total_loss /= len(train_dataloader)
    print(f"Epoch {epoch+1}/{EPOCHS} Loss: {total_loss}")
```

```python
torch_rnn.eval()
correct = 0
total = 0

with torch.no_grad():
    for x_batch, y_batch in test_dataloader:
        x_batch = x_batch.to(device)
        y_batch = y_batch.to(device)

        outputs = torch_rnn(x_batch).squeeze(1)
        predictions = torch.sigmoid(outputs) >= 0.5

        correct += (predictions == y_batch.bool()).sum().item()
        total += y_batch.size(0)

accuracy = correct / total
print(f"Test Accuracy: {accuracy:.4f}")
```
